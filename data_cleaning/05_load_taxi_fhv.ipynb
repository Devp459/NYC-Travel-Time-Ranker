```python
S = "workspace.default"
base = "/Volumes/workspace/default/nyc_inputs"

for name in ["yellow_clean", "fhv_clean"]:
    df = spark.read.option("header", True).option("inferSchema", True).csv(f"{base}/{name}.csv")
    df.write.mode("overwrite").saveAsTable(f"{S}.{name}")
    print(name, spark.table(f"{S}.{name}").count())
    df.printSchema()
```

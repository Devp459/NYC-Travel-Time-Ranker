```python
import zipfile, glob, os, tempfile

src = "/Volumes/workspace/default/nyc_inputs"
dst = "/Volumes/workspace/default/citibike"

def copy_out(zf, name, path):
    with zf.open(name) as f, open(path, "wb") as out:
        while True:
            chunk = f.read(1 << 24)
            if not chunk:
                break
            out.write(chunk)

def extract(zf):
    for name in zf.namelist():
        base = os.path.basename(name)
        if name.startswith("__MACOSX") or not base or base.startswith("."):
            continue
        if base.endswith(".zip"):               # a zip inside the zip
            tmp = os.path.join(tempfile.mkdtemp(), base)
            copy_out(zf, name, tmp)
            with zipfile.ZipFile(tmp) as inner:
                extract(inner)
            os.remove(tmp)
        elif base.endswith(".csv") and not base.startswith("JC-"):
            copy_out(zf, name, f"{dst}/{base}")
            print("wrote", base)

for p in sorted(glob.glob(f"{src}/*citibike*.zip")):
    print("==", os.path.basename(p))
    with zipfile.ZipFile(p) as zf:
        extract(zf)

print(sorted(os.listdir(dst)))
```
